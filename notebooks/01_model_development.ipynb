```python

from pathlib import Path
import pandas as pd

DATA_PATH = Path("../data/raw/irrigation_prediction.csv")

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nTarget distribution:")
print(df["Irrigation_Need"].value_counts(dropna=False))
```

```python

FEATURES = [
    "Soil_Moisture",
    "Crop_Type",
    "Crop_Growth_Stage",
    "Soil_Type",
    "Temperature_C",
    "Humidity",
]

TARGET = "Irrigation_Need"

X = df[FEATURES].copy()
y = df[TARGET].copy()

print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeature data types:")
print(X.dtypes)

print("\nTarget classes:")
print(sorted(y.unique()))

print("\nMissing values in X:", X.isnull().sum().sum())
print("Missing values in y:", y.isnull().sum())
```

```python

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

NUMERIC_FEATURES = [
    "Soil_Moisture",
    "Temperature_C",
    "Humidity",
]

CATEGORICAL_FEATURES = [
    "Crop_Type",
    "Crop_Growth_Stage",
    "Soil_Type",
]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y,
)

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore")),
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, NUMERIC_FEATURES),
    ("categorical", categorical_pipeline, CATEGORICAL_FEATURES),
])

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Training rows:", X_train.shape[0])
print("Testing rows:", X_test.shape[0])
print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    f1_score,
)

baseline_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        class_weight="balanced",
        max_iter=2000,
        random_state=42,
    )),
])

baseline_model.fit(X_train, y_train)

y_pred = baseline_model.predict(X_test)

print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("Macro F1:", round(f1_score(y_test, y_pred, average="macro"), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    labels=["Low", "Medium", "High"],
    zero_division=0,
))

print("Confusion Matrix (rows = actual, columns = predicted):")
print(confusion_matrix(
    y_test,
    y_pred,
    labels=["Low", "Medium", "High"],
))
```

```python

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, f1_score, accuracy_score

rf_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=300,
        min_samples_leaf=3,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1,
    )),
])

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)

print("Random Forest Accuracy:", round(accuracy_score(y_test, rf_pred), 4))
print("Random Forest Macro F1:", round(f1_score(y_test, rf_pred, average="macro"), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    rf_pred,
    labels=["Low", "Medium", "High"],
    zero_division=0,
))
```

```python

from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42,
)

scoring = {
    "accuracy": "accuracy",
    "macro_f1": "f1_macro",
}

cv_results = cross_validate(
    rf_model,
    X_train,
    y_train,
    cv=cv,
    scoring=scoring,
    n_jobs=-1,
)

print("5-Fold Cross-Validation Results")
print(
    "Accuracy: "
    f"{cv_results['test_accuracy'].mean():.4f} "
    f"+/- {cv_results['test_accuracy'].std():.4f}"
)
print(
    "Macro F1: "
    f"{cv_results['test_macro_f1'].mean():.4f} "
    f"+/- {cv_results['test_macro_f1'].std():.4f}"
)
```

```python

from xgboost import XGBClassifier
from sklearn.pipeline import Pipeline
from sklearn.model_selection import cross_validate
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
)

# XGBoost requires class labels encoded as integers
class_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2,
}

y_train_encoded = y_train.map(class_mapping)
y_test_encoded = y_test.map(class_mapping)

xgb_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", XGBClassifier(
        objective="multi:softprob",
        num_class=3,
        n_estimators=300,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="mlogloss",
        random_state=42,
        n_jobs=-1,
    )),
])

scoring = {
    "accuracy": "accuracy",
    "macro_f1": "f1_macro",
}

xgb_cv = cross_validate(
    xgb_model,
    X_train,
    y_train_encoded,
    cv=cv,
    scoring=scoring,
    n_jobs=1,
)

print("XGBoost 5-Fold CV")
print(
    f"Accuracy: {xgb_cv['test_accuracy'].mean():.4f} "
    f"+/- {xgb_cv['test_accuracy'].std():.4f}"
)
print(
    f"Macro F1: {xgb_cv['test_macro_f1'].mean():.4f} "
    f"+/- {xgb_cv['test_macro_f1'].std():.4f}"
)

# Fit on all training data, then evaluate on the held-out test set
xgb_model.fit(X_train, y_train_encoded)
xgb_pred = xgb_model.predict(X_test)

print("\nHeld-out Test Results")
print("Accuracy:", round(accuracy_score(y_test_encoded, xgb_pred), 4))
print("Macro F1:", round(f1_score(y_test_encoded, xgb_pred, average="macro"), 4))

print("\nClassification Report:")
print(classification_report(
    y_test_encoded,
    xgb_pred,
    labels=[0, 1, 2],
    target_names=["Low", "Medium", "High"],
    zero_division=0,
))
```

```python

import numpy as np
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.metrics import classification_report, f1_score, accuracy_score

# Calculate weights using training labels only
sample_weights = compute_sample_weight(
    class_weight="balanced",
    y=y_train_encoded,
)

# Reuse the same XGBoost configuration
xgb_balanced = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", XGBClassifier(
        objective="multi:softprob",
        num_class=3,
        n_estimators=300,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="mlogloss",
        random_state=42,
        n_jobs=-1,
    )),
])

# Fit using sample weights
xgb_balanced.fit(
    X_train,
    y_train_encoded,
    classifier__sample_weight=sample_weights,
)

balanced_pred = xgb_balanced.predict(X_test)

print("Balanced XGBoost Accuracy:",
      round(accuracy_score(y_test_encoded, balanced_pred), 4))

print("Balanced XGBoost Macro F1:",
      round(f1_score(y_test_encoded, balanced_pred, average="macro"), 4))

print("\nClassification Report:")
print(classification_report(
    y_test_encoded,
    balanced_pred,
    labels=[0, 1, 2],
    target_names=["Low", "Medium", "High"],
    zero_division=0,
))
```

```python

from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

final_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=300,
        min_samples_leaf=3,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1,
    )),
])

# Fit only on the training split; keep the test split untouched
final_model.fit(X_train, y_train)

print("Final candidate fitted successfully.")
print("Model classes:", final_model.named_steps["classifier"].classes_)
```

```python

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix,
)

final_pred = final_model.predict(X_test)

print("Final Random Forest — Test Results")
print("Accuracy:", round(accuracy_score(y_test, final_pred), 4))
print("Macro F1:", round(f1_score(y_test, final_pred, average="macro"), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    final_pred,
    labels=["Low", "Medium", "High"],
    zero_division=0,
))

print("Confusion Matrix (rows = actual, columns = predicted):")
print(confusion_matrix(
    y_test,
    final_pred,
    labels=["Low", "Medium", "High"],
))
```
